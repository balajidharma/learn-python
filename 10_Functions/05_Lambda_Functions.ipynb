# Lambda Functions

A `lambda` is a small, anonymous, single-expression function. It's created with the `lambda` keyword instead of `def`, has no name of its own, and automatically returns the value of its expression.

**Syntax:** `lambda arguments: expression`

## 1. Basic lambda vs a regular function

A lambda is equivalent to a tiny one-line function, without the `def` and `return`.

```python
def square_def(x):
    return x * x

square_lambda = lambda x: x * x

print(square_def(5))
print(square_lambda(5))
```

## 2. Multiple arguments

```python
add = lambda a, b: a + b
print(add(3, 4))

greet = lambda name, greeting='Hello': f'{greeting}, {name}!'
print(greet('Alice'))
print(greet('Bob', 'Hi'))
```

## 3. Using lambda as a sort key

The most common real-world use: telling `sorted()` or `.sort()` what to sort by, without writing a whole named function.

```python
people = [('Alice', 30), ('Bob', 25), ('Charlie', 35)]

by_age = sorted(people, key=lambda person: person[1])
print(by_age)

words = ['banana', 'kiwi', 'apple', 'fig']
by_length = sorted(words, key=lambda w: len(w))
print(by_length)
```

## 4. lambda with map() and filter()

`map()` applies a function to every item in an iterable; `filter()` keeps only the items where the function returns `True`.

```python
numbers = [1, 2, 3, 4, 5, 6]

squared = list(map(lambda n: n ** 2, numbers))
print(squared)

evens = list(filter(lambda n: n % 2 == 0, numbers))
print(evens)
```

## 5. lambda inside another function

Lambdas are often built on the fly and returned or passed around, similar to closures.

```python
def make_adder(n):
    return lambda x: x + n

add_5 = make_adder(5)
add_10 = make_adder(10)

print(add_5(3))
print(add_10(3))
```

## 6. When NOT to use lambda

Lambdas can only contain a single expression — no statements, no multiple lines, no assignments. For anything beyond a simple expression, use a regular `def` function; it's more readable and can have a docstring.

```python
# OK: simple expression
discount_price = lambda price, pct: price * (1 - pct)
print(discount_price(100, 0.2))

# Avoid: cramming complex logic into a lambda hurts readability.
# Prefer a named function instead:
def discount_price_readable(price, pct):
    """Apply a percentage discount to a price."""
    if not 0 <= pct <= 1:
        raise ValueError('pct must be between 0 and 1')
    return price * (1 - pct)

print(discount_price_readable(100, 0.2))
```
