# Return Values

The `return` statement sends a value back to the caller and immediately ends the function. A function that never hits `return` (or uses a bare `return`) implicitly returns `None`.

## 1. Basic return

```python
def add(a, b):
    return a + b

result = add(3, 4)
print(result)
```

## 2. Functions without return -> None

```python
def greet(name):
    print(f'Hello, {name}!')   # prints, but does not return anything

result = greet('Alice')
print(result)
print(type(result))
```

## 3. return vs print

`print()` only displays a value — it doesn't give the caller anything to work with. `return` hands a value back so it can be stored, passed along, or used in further calculations.

```python
def add_print(a, b):
    print(a + b)     # just displays it

def add_return(a, b):
    return a + b      # hands the value back

x = add_print(2, 3)     # prints 5, but x is None
print('x =', x)

y = add_return(2, 3)    # nothing printed here
print('y =', y)
print('y doubled =', y * 2)   # we can keep using the returned value
```

## 4. Returning multiple values

Python lets you return several values separated by commas — they're packed into a tuple, which you can unpack directly.

```python
def min_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = min_max([4, 9, 1, 7, 3])
print(f'lowest={lowest}, highest={highest}')

# Or keep it as a tuple
result = min_max([4, 9, 1, 7, 3])
print(result, type(result))
```

## 5. Early return / guard clauses

`return` exits the function immediately, which is useful for handling edge cases up front instead of nesting the main logic inside an `if`.

```python
def safe_divide(a, b):
    if b == 0:
        return None   # exit early, avoid the error
    return a / b

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

## 6. Returning different types conditionally

A function can return different types of values along different code paths — Python doesn't enforce a single return type.

```python
def parse_number(text):
    try:
        return int(text)
    except ValueError:
        return None

print(parse_number('42'))
print(parse_number('abc'))
```

## 7. Chaining function calls

Since a function call is an expression, its return value can be fed directly into another call.

```python
def square(x):
    return x * x

def increment(x):
    return x + 1

print(square(increment(4)))   # increment(4) -> 5, then square(5) -> 25
```
