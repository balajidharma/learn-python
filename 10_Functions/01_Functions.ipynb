# Python Functions — Overview

A function is a reusable, named block of code that performs a task. Functions let you avoid repeating yourself, break a problem into smaller pieces, and give a name to a piece of logic so it's easier to read and test.

```python
def greet(name):
    return f'Hello, {name}!'

print(greet('World'))
```
