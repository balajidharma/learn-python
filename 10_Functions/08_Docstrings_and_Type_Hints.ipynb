# Docstrings and Type Hints

Two tools make functions easier to understand and safer to use, without changing how they run:

- **Docstrings** — a string literal right after `def` that documents what the function does
- **Type hints** — optional annotations that describe the expected types of arguments and return values

## 1. Writing a docstring

```python
def add(a, b):
    """Return the sum of a and b."""
    return a + b

print(add(2, 3))
print(add.__doc__)
```

## 2. Multi-line docstrings

For anything beyond a one-liner, describe the parameters, return value, and any exceptions raised.

```python
def divide(a, b):
    """Divide a by b.

    Args:
        a (float): The numerator.
        b (float): The denominator. Must not be zero.

    Returns:
        float: The result of a / b.

    Raises:
        ZeroDivisionError: If b is 0.
    """
    return a / b

print(divide(10, 2))
print(divide.__doc__)
```

## 3. Viewing docs with help()

`help()` displays a function's signature and docstring — handy in an interactive session without reading the source code.

```python
def divide(a, b):
    """Divide a by b.

    Args:
        a (float): The numerator.
        b (float): The denominator. Must not be zero.

    Returns:
        float: The result of a / b.

    Raises:
        ZeroDivisionError: If b is 0.
    """
    return a / b

help(divide)
```

## 4. Type hints (annotations)

Add `: type` after a parameter and `-> type` before the colon to document expected types. Python does **not** enforce these at runtime — they're for readability and for tools like editors, linters, and `mypy`.

```python
def greet(name: str, times: int = 1) -> str:
    """Return a greeting repeated `times` times."""
    return f'Hello, {name}! ' * times

print(greet('Alice'))
print(greet('Bob', 3))

# Hints are NOT enforced -- this still runs, even though 42 isn't a str
print(greet(42))
```

## 5. Inspecting type hints

Annotations are stored on the function and can be read at runtime via `__annotations__`.

```python
def add(a: int, b: int) -> int:
    return a + b

print(add.__annotations__)
```

## 6. Hinting more complex types

The `typing` module (or built-in generics on modern Python) lets you describe lists, optional values, and multiple possible types.

```python
from typing import Optional, Union

def find_user(user_id: int) -> Optional[str]:
    """Return the username for user_id, or None if not found."""
    users = {1: 'alice', 2: 'bob'}
    return users.get(user_id)

print(find_user(1))
print(find_user(99))

def process(value: Union[int, str]) -> str:
    """Accept either an int or a str and return it as a string."""
    return str(value)

print(process(42))
print(process('hello'))

# Modern shorthand (Python 3.10+) for lists and unions:
def total(numbers: list[int]) -> int:
    return sum(numbers)

print(total([1, 2, 3]))
```

## 7. Why bother with docstrings and hints?

- **Docstrings** make `help()`, IDE tooltips, and auto-generated documentation useful.
- **Type hints** help catch bugs early with static checkers (like `mypy`), and make function signatures self-explanatory.
- Neither changes runtime behavior — Python remains dynamically typed either way.
