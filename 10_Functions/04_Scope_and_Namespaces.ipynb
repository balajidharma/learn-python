# Scope and Namespaces

Scope determines where in your code a variable name can be seen and used. Python looks up names using the **LEGB** rule:

| Scope | Meaning |
|-------|---------|
| **L**ocal | Names assigned inside the current function |
| **E**nclosing | Names in any enclosing (outer) function |
| **G**lobal | Names assigned at the top level of the module |
| **B**uilt-in | Names Python provides automatically (`len`, `print`, ...) |

## 1. Local scope

A variable created inside a function only exists inside that function.

```python
def my_function():
    message = 'I am local'
    print(message)

my_function()

try:
    print(message)   # NameError: message doesn't exist out here
except NameError as e:
    print(f'NameError: {e}')
```

## 2. Global scope

A variable created at the top level of a script is global — functions can **read** it without any special keyword.

```python
counter = 0   # global

def show_counter():
    print(f'counter is {counter}')   # reading a global is fine

show_counter()
```

## 3. Reassigning a global from inside a function

Assigning to a name inside a function makes it **local by default** — even if a global with the same name exists. Use the `global` keyword to explicitly modify the global variable.

```python
counter = 0

def increment_broken():
    counter += 1   # UnboundLocalError: treated as local because we assign to it

try:
    increment_broken()
except UnboundLocalError as e:
    print(f'UnboundLocalError: {e}')

def increment_fixed():
    global counter
    counter += 1

increment_fixed()
increment_fixed()
print(f'counter is now {counter}')
```

## 4. Enclosing scope and closures

A nested function can **read** variables from its enclosing function. This is the basis of closures — a function that 'remembers' values from where it was created.

```python
def make_multiplier(factor):
    def multiply(x):
        return x * factor   # 'factor' comes from the enclosing scope
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))
print(triple(5))
```

## 5. nonlocal — modifying an enclosing variable

Just like `global` lets you modify a global from inside a function, `nonlocal` lets a nested function modify a variable from its enclosing (but not global) scope.

```python
def make_counter():
    count = 0
    def increment():
        nonlocal count
        count += 1
        return count
    return increment

counter = make_counter()
print(counter())
print(counter())
print(counter())
```

## 6. Built-in scope

Names like `print`, `len`, and `sum` live in the built-in scope — always available unless you shadow them with your own variable of the same name (best avoided!).

```python
print(len([1, 2, 3]))

# Shadowing a built-in (avoid doing this in real code!)
len = 'oops, len is now a string'
try:
    print(len([1, 2, 3]))
except TypeError as e:
    print(f'TypeError: {e}')
del len   # restore the built-in
print(len([1, 2, 3]))
```
