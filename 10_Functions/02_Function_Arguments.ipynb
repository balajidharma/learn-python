# Function Arguments

A function is defined with `def`, and can accept **parameters** — placeholders for the values (**arguments**) the caller passes in. Python offers several flexible ways to pass arguments.

**Key Points:**
- Positional arguments are matched by order
- Keyword arguments are matched by name
- Default values make a parameter optional
- `*args` collects extra positional arguments; `**kwargs` collects extra keyword arguments

## 1. Defining and calling a basic function

```python
def greet(name):
    print(f'Hello, {name}!')

greet('Alice')
greet('Bob')
```

## 2. Positional arguments

Arguments are matched to parameters **by position**, in the order they're passed.

```python
def describe_pet(animal, name):
    print(f'{name} is a {animal}')

describe_pet('dog', 'Rex')      # animal='dog', name='Rex'
describe_pet('Rex', 'dog')      # order matters! output is reversed
```

## 3. Keyword arguments

Passing `name=value` lets you specify arguments by name, in any order, which makes the call self-documenting.

```python
def describe_pet(animal, name):
    print(f'{name} is a {animal}')

describe_pet(animal='dog', name='Rex')
describe_pet(name='Whiskers', animal='cat')   # order doesn't matter with keywords
```

## 4. Default parameter values

A default value makes an argument optional. If the caller doesn't supply it, the default is used.

```python
def greet(name, greeting='Hello'):
    print(f'{greeting}, {name}!')

greet('Alice')
greet('Bob', 'Hi')
greet('Charlie', greeting='Welcome')
```

## 5. *args — variable positional arguments

Collects any number of extra positional arguments into a tuple.

```python
def total(*numbers):
    print(numbers, type(numbers))
    return sum(numbers)

print(total(1, 2, 3))
print(total(10, 20, 30, 40))
```

## 6. **kwargs — variable keyword arguments

Collects any number of extra keyword arguments into a dictionary.

```python
def build_profile(**details):
    print(details, type(details))
    for key, value in details.items():
        print(f'  {key}: {value}')

build_profile(name='Alice', age=30, city='NYC')
```

## 7. Combining all argument types

When mixing styles, the order in a function definition must be:
`def f(positional, default=value, *args, **kwargs)`

```python
def order_summary(customer, item, quantity=1, *extras, **details):
    print(f'Customer: {customer}')
    print(f'Item: {item} x{quantity}')
    if extras:
        print(f'Extras: {extras}')
    if details:
        print(f'Details: {details}')

order_summary('Alice', 'Book', 2, 'giftwrap', 'express', discount='10%')
```

## 8. Keyword-only arguments

Parameters listed after a bare `*` **must** be passed by keyword — this avoids ambiguous calls for functions with many boolean-ish flags.

```python
def create_user(username, *, is_admin=False, is_active=True):
    print(f'{username}: admin={is_admin}, active={is_active}')

create_user('alice', is_admin=True)

try:
    create_user('bob', True)   # error: is_admin must be passed by keyword
except TypeError as e:
    print(f'TypeError: {e}')
```

## 9. Positional-only parameters

Parameters listed before a bare `/` **must** be passed positionally — useful when the parameter name isn't meaningful to the caller.

```python
def power(base, exponent, /):
    return base ** exponent

print(power(2, 10))

try:
    power(base=2, exponent=10)   # error: these are positional-only
except TypeError as e:
    print(f'TypeError: {e}')
```

## 10. Mutable default arguments — a common pitfall

A mutable default (like a list or dict) is created **once**, when the function is defined — not on every call. Reusing it across calls can cause surprising bugs.

```python
def add_item(item, basket=[]):   # BUG: default list is shared across calls
    basket.append(item)
    return basket

print(add_item('apple'))
print(add_item('banana'))   # 'apple' is still there!

# Fix: use None as the default, create a new list inside the function
def add_item_fixed(item, basket=None):
    if basket is None:
        basket = []
    basket.append(item)
    return basket

print(add_item_fixed('apple'))
print(add_item_fixed('banana'))
```
