# Decorators

In Python, functions are **first-class objects** — they can be assigned to variables, passed as arguments, and returned from other functions. A **decorator** is a function that takes another function, wraps it with extra behavior, and returns the wrapped version — without changing the original function's code.

## 1. Functions are first-class objects

```python
def shout(text):
    return text.upper()

# Assign a function to a variable
yell = shout
print(yell('hello'))

# Pass a function as an argument
def apply(func, value):
    return func(value)

print(apply(shout, 'hi there'))
```

## 2. A function that returns a function

```python
def outer():
    def inner():
        return 'Hello from inner!'
    return inner

fn = outer()
print(fn())
```

## 3. Writing a basic decorator

A decorator wraps a function inside another function, typically running code **before** and/or **after** the original call.

```python
def my_decorator(func):
    def wrapper():
        print('Before the function runs')
        func()
        print('After the function runs')
    return wrapper

def say_hello():
    print('Hello!')

say_hello = my_decorator(say_hello)   # manually decorating
say_hello()
```

## 4. The @ syntax

`@my_decorator` above a function definition is shorthand for `func = my_decorator(func)`.

```python
def my_decorator(func):
    def wrapper():
        print('Before the function runs')
        func()
        print('After the function runs')
    return wrapper

@my_decorator
def say_hello():
    print('Hello!')

say_hello()
```

## 5. Decorating functions with arguments

Use `*args` and `**kwargs` in the wrapper so the decorator works with **any** function signature.

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f'Calling {func.__name__} with args={args}, kwargs={kwargs}')
        result = func(*args, **kwargs)
        print(f'{func.__name__} returned {result}')
        return result
    return wrapper

@logger
def add(a, b):
    return a + b

add(3, 4)
```

## 6. Preserving function metadata with functools.wraps

Without `@functools.wraps`, the decorated function loses its original name and docstring, which can confuse debugging and documentation tools.

```python
import functools

def logger(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@logger
def add(a, b):
    """Add two numbers together."""
    return a + b

print(add.__name__)   # 'add', not 'wrapper'
print(add.__doc__)
```

## 7. A practical example: timing a function

```python
import time
import functools

def timer(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = time.perf_counter() - start
        print(f'{func.__name__} took {elapsed:.6f} seconds')
        return result
    return wrapper

@timer
def slow_sum(n):
    return sum(range(n))

print(slow_sum(1_000_000))
```

## 8. Stacking multiple decorators

Decorators apply bottom-up: the one closest to the function runs first, then wraps outward.

```python
def bold(func):
    def wrapper(*args, **kwargs):
        return f'<b>{func(*args, **kwargs)}</b>'
    return wrapper

def italic(func):
    def wrapper(*args, **kwargs):
        return f'<i>{func(*args, **kwargs)}</i>'
    return wrapper

@bold
@italic
def greet(name):
    return f'Hello, {name}!'

print(greet('Alice'))   # italic applied first, then bold wraps that
```
