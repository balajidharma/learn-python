# Setting Up Python Environment
To start coding in Python, you need to set up your environment:
1. **Install Python:** Download from [python.org](https://www.python.org/downloads/).
2. **Choose an IDE or Editor:** Popular options include VS Code, PyCharm, Jupyter Notebook, or even a simple text editor.
3. **Verify Installation:** Open a terminal and run:
```python
python --version
```
4. **(Optional) Set up a Virtual Environment:**
```python
python -m venv myenv
source myenv/bin/activate  # On Windows: myenv\Scripts\activate
```
5. **Install Useful Packages:**
```python
pip install numpy pandas matplotlib jupyter
```
