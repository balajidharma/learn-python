# Match Statement (Python 3.10+)

Pattern matching — more powerful than switch/case. Matches values, types, sequences, and dicts.

**Key Points:**
- `case _` is the default catch-all
- `|` matches multiple patterns
- Guards add extra `if` conditions
- Can destructure sequences and dicts

## 1. Basic value matching

```python
def http_status(code):
    match code:
        case 200: return 'OK'
        case 201: return 'Created'
        case 400: return 'Bad Request'
        case 404: return 'Not Found'
        case 500: return 'Internal Server Error'
        case _:   return 'Unknown'

for code in [200, 404, 500, 999]:
    print(f"{code} -> {http_status(code)}")
```

## 2. OR patterns with |

```python
def classify_day(day):
    match day.lower():
        case 'saturday' | 'sunday': return 'Weekend'
        case 'monday' | 'tuesday' | 'wednesday' | 'thursday' | 'friday': return 'Weekday'
        case _: return 'Invalid'

for day in ['Monday','Saturday','Wednesday','Sunday']:
    print(f"{day} -> {classify_day(day)}")
```

## 3. Guards (if inside case)

```python
def classify_number(n):
    match n:
        case 0: return 'Zero'
        case n if n < 0: return f'Negative ({n})'
        case n if n % 2 == 0: return f'Positive even ({n})'
        case _: return f'Positive odd ({n})'

for n in [0, -4, 6, 9]:
    print(classify_number(n))
```

## 4. Matching sequences

```python
def describe_point(point):
    match point:
        case (0, 0):  return 'Origin'
        case (x, 0):  return f'On x-axis at {x}'
        case (0, y):  return f'On y-axis at {y}'
        case (x, y):  return f'Point at ({x}, {y})'

for p in [(0,0),(3,0),(0,5),(2,4)]:
    print(f"{p} -> {describe_point(p)}")
```

## 5. Matching dictionaries

```python
def handle_command(cmd):
    match cmd:
        case {'action': 'quit'}: return 'Quitting...'
        case {'action': 'move', 'direction': d}: return f'Moving {d}'
        case {'action': 'attack', 'weapon': w}: return f'Attacking with {w}'
        case _: return 'Unknown command'

for cmd in [{'action':'quit'},{'action':'move','direction':'north'},{'action':'attack','weapon':'sword'},{'action':'dance'}]:
    print(handle_command(cmd))
```
