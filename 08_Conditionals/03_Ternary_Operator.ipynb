# Ternary Operator

A one-line shorthand for a simple `if / else`.

**Syntax:** `value_if_true if condition else value_if_false`

## 1. Basic ternary

```python
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)
```

## 2. In f-strings

```python
items = 1
print(f'You have {items} item{"s" if items != 1 else ""} in your cart')
items = 5
print(f'You have {items} item{"s" if items != 1 else ""} in your cart')
```

## 3. In list comprehensions

```python
numbers = [1, -3, 7, -2, 5]
absolutes = [n if n >= 0 else -n for n in numbers]
print(absolutes)

labels = ['positive' if n > 0 else 'negative' if n < 0 else 'zero' for n in numbers]
print(labels)
```

## 4. When NOT to use ternary

Avoid it for complex logic — use `if/elif/else` instead for readability.

```python
score = 85
if score >= 90:   grade = 'A'
elif score >= 80: grade = 'B'
elif score >= 70: grade = 'C'
elif score >= 60: grade = 'D'
else:             grade = 'F'
print(grade)
```
