# Break, Continue, Pass

| Keyword | What it does |
|---------|-------------|
| `break` | Exit the loop immediately |
| `continue` | Skip rest of this iteration |
| `pass` | Do nothing — placeholder |

## 1. break

```python
numbers = [3, 7, 1, -4, 9, 2]
for n in numbers:
    if n < 0:
        print(f'Found negative: {n}, stopping.')
        break
    print(n)
```

## 2. continue

```python
for n in range(1, 11):
    if n % 2 == 0:
        continue
    print(n, end=' ')
print()

words = ['hello', '', 'world', '', 'python']
for w in words:
    if not w:
        continue
    print(w)
```

## 3. pass

```python
for n in range(1, 6):
    if n == 3:
        pass  # TODO: handle later
    else:
        print(n, end=' ')
print()

def placeholder():
    pass

class EmptyClass:
    pass

print('No error — pass is silent')
```

## 4. Combining break and continue

```python
scores = [88, 92, -1, 76, 55]
total = count = 0
for score in scores:
    if score == -1:
        print('Sentinel found — stopping')
        break
    if score < 0:
        print(f'Skipping invalid: {score}')
        continue
    total += score
    count += 1
if count:
    print(f'Average of {count} scores: {total/count:.1f}')
```
