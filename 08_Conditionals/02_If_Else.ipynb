# If / Elif / Else

The `if` statement is the most fundamental way to make decisions in Python.

**Key Points:**
- `if` runs when condition is True
- `elif` checked only if prior conditions were False
- `else` runs when no condition matched
- Combine with `and`, `or`, `not`

## 1. Basic if

```python
temperature = 35

if temperature > 30:
    print("It's hot outside!")
```

## 2. if / else

```python
age = 16

if age >= 18:
    print("You can vote")
else:
    print("Too young to vote")
```

## 3. if / elif / else

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```

## 4. Nested if

```python
is_member = True
purchase = 120

if is_member:
    if purchase >= 100:
        print("20% discount applied")
    else:
        print("10% member discount")
else:
    print("No discount")
```

## 5. Logical operators

```python
age = 25
has_license = True
if age >= 18 and has_license:
    print("Can drive")

is_holiday = True
if is_holiday or False:
    print("Day off!")

is_raining = False
if not is_raining:
    print("Good weather")
```

## 6. Truthy and Falsy values

**Falsy:** `0`, `0.0`, `""`, `[]`, `{}`, `None`, `False`  
**Truthy:** everything else

```python
values = [0, 1, '', 'hello', [], [1,2], None, True, False]
for v in values:
    print(f"{repr(v):<12} → {'Truthy' if v else 'Falsy'}")
```
