# Comprehensions

Concise, Pythonic way to create collections from iterables in a single readable line.

| Type | Syntax |
|------|--------|
| List | `[expr for item in iterable if cond]` |
| Dict | `{k: v for item in iterable if cond}` |
| Set  | `{expr for item in iterable if cond}` |
| Generator | `(expr for item in iterable if cond)` |

## 1. List comprehension

```python
squares = [x**2 for x in range(1, 6)]
print(squares)

evens = [x for x in range(1, 21) if x % 2 == 0]
print(evens)

words = ['hello', 'world', 'python']
print([w.capitalize() for w in words])
```

## 2. With condition

```python
numbers = range(-5, 6)
positive_squares = [x**2 for x in numbers if x > 0]
print(positive_squares)

sentence = 'The quick brown fox jumps over the lazy dog'
long_words = [w for w in sentence.split() if len(w) > 3]
print(long_words)
```

## 3. Nested comprehension

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
flat = [n for row in matrix for n in row]
print(flat)

transposed = [[row[i] for row in matrix] for i in range(3)]
for row in transposed:
    print(row)
```

## 4. Dict comprehension

```python
squares = {x: x**2 for x in range(1, 6)}
print(squares)

scores = {'Alice':92,'Bob':74,'Charlie':85,'Diana':68}
passed = {n: s for n, s in scores.items() if s >= 80}
print(passed)

original = {'a':1,'b':2,'c':3}
print({v: k for k, v in original.items()})
```

## 5. Set comprehension

```python
words = ['cat','dog','elephant','ant','ox','bee']
lengths = {len(w) for w in words}
print(sorted(lengths))

numbers = [1,2,2,3,3,3,4]
print({x**2 for x in numbers})
```

## 6. Generator expressions

Like list comprehensions but lazy — generate values one at a time. Use `()` instead of `[]`.

```python
squares_list = [x**2 for x in range(5)]
print('List:', squares_list)

squares_gen = (x**2 for x in range(5))
print('Generator:', squares_gen)
print('next():', next(squares_gen))
print('next():', next(squares_gen))
print('sum 1-5:', sum(x**2 for x in range(1, 6)))
```
