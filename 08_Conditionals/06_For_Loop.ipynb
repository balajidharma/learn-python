# For Loop

Iterates over any sequence or iterable. Use when you know what you're looping over.

**Key Points:**
- Works with lists, strings, ranges, dicts, tuples, sets
- `range()` generates number sequences
- `enumerate()` gives index + value
- `zip()` pairs multiple iterables

## 1. Iterating over a list

```python
fruits = ['apple', 'banana', 'cherry', 'mango']
for fruit in fruits:
    print(fruit)
```

## 2. Iterating over a string

```python
for char in 'Python':
    print(char, end=' ')
print()

vowels = sum(1 for c in 'Learn Python' if c.lower() in 'aeiou')
print(f'Vowels: {vowels}')
```

## 3. range()

```python
print('range(5):     ', list(range(5)))
print('range(2,8):   ', list(range(2, 8)))
print('range(0,20,4):', list(range(0, 20, 4)))
print('Countdown:    ', list(range(10, 0, -2)))
```

## 4. enumerate()

```python
langs = ['Python', 'JavaScript', 'Rust', 'Go']
for i, lang in enumerate(langs, start=1):
    print(f'{i}. {lang}')
```

## 5. zip()

```python
names  = ['Alice', 'Bob', 'Charlie']
scores = [92, 85, 78]
grades = ['A', 'B', 'C']
for name, score, grade in zip(names, scores, grades):
    print(f'{name:<10} {score}  {grade}')
```

## 6. Iterating over a dictionary

```python
person = {'name':'Alice','age':30,'city':'NYC'}
for key, value in person.items():
    print(f'  {key}: {value}')
```

## 7. Nested for loops

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(f'{i}x{j}={i*j}', end='  ')
    print()
```

## 8. for / else

```python
for n in range(2, 15):
    for i in range(2, n):
        if n % i == 0:
            break
    else:
        print(n, end=' ')
print()
```
