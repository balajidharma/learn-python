# While Loop

Repeats a block **while** a condition is True. Use when you don't know how many iterations you need.

**Key Points:**
- Always ensure the condition eventually becomes False
- Use `break` to exit early
- Use `continue` to skip to next iteration
- `else` runs when loop finishes without a `break`

## 1. Basic while

```python
count = 1
while count <= 5:
    print(count, end=' ')
    count += 1
print()
```

## 2. while with break

```python
n = 1
while True:
    if n % 3 == 0 and n % 7 == 0:
        print(f"First divisible by 3 and 7: {n}")
        break
    n += 1
```

## 3. while with continue

```python
n = 0
while n < 10:
    n += 1
    if n % 2 == 0:
        continue
    print(n, end=' ')
print()
```

## 4. while / else

```python
attempts = 0
guesses = [1111, 2222, 1234]
correct = 1234

while attempts < len(guesses):
    pin = guesses[attempts]
    attempts += 1
    if pin == correct:
        print(f"Access granted on attempt {attempts}!")
        break
    print(f"Wrong PIN: {pin}")
else:
    print("All attempts failed")
```

## 5. Infinite loop with controlled exit

```python
choices = ['1', '2', 'q']
idx = 0
while True:
    choice = choices[idx]; idx += 1
    print(f"User chose: {choice}")
    if choice == 'q':
        print('Goodbye!')
        break
    elif choice == '1':
        print('  -> Option 1')
    elif choice == '2':
        print('  -> Option 2')
```
