# Variables


Variables are used to store data in a program. In Python, you do not need to declare the type of a variable. The type is inferred from the value assigned.

```python
# Example 1: Assigning Values to Variables
x = 5           # Integer
y = 3.14        # Float
name = "Alice"   # String
is_valid = True # Boolean

print(x)
print(y)
print(name)
print(is_valid)
```

```python
# Example 2: Changing Variable Values
x = 10
name = "Bob"
print(x)
print(name)
```

```python
# Example 3: Multiple Assignments
a, b, c = 1, 2, 3
x = y = z = 0
print(a, b, c)
print(x, y, z)
```

```python
# Example 4: Variable Naming Rules
# Valid variable names
my_var = 1
_var2 = 2
var_3 = 3

# Invalid variable names (uncomment to see errors)
# 2var = 4
# my-var = 5
# for = 6  # 'for' is a reserved keyword

print(my_var, _var2, var_3)
```
