# Python Data Types

Python has several built-in data types that are used to store different kinds of values. The most common data types are:

- **int**: Integer numbers
- **float**: Decimal numbers
- **str**: Text (string)
- **bool**: Boolean values (True or False)
- **NoneType**: Represents the absence of a value

Below are examples of each data type.

```python
# Integer (int)
num = 10
print(num, type(num))
```

```python
# Float (float)
pi = 3.14159
print(pi, type(pi))
```

```python
# String (str)
message = "Hello, Python!"
print(message, type(message))
```

```python
# Boolean (bool)
is_active = True
print(is_active, type(is_active))
```

```python
# NoneType
nothing = None
print(nothing, type(nothing))
```
