# Bitwise Operators

Bitwise operators work directly on the binary representation of integers, one bit at a time. They're commonly used in low-level programming, flags/permissions, and performance-sensitive code.

| Operator | Name | Example |
|----------|------|---------|
| `&` | AND | `5 & 3` |
| `\|` | OR | `5 \| 3` |
| `^` | XOR | `5 ^ 3` |
| `~` | NOT (bitwise complement) | `~5` |
| `<<` | Left shift | `5 << 1` |
| `>>` | Right shift | `5 >> 1` |

## 1. AND (&), OR (|), XOR (^)

`5` is `0101` in binary and `3` is `0011`.

```python
a = 5   # 0101
b = 3   # 0011

print(a & b, bin(a & b))   # AND: 1 only where both bits are 1
print(a | b, bin(a | b))   # OR: 1 where either bit is 1
print(a ^ b, bin(a ^ b))   # XOR: 1 where bits differ
```

## 2. NOT (~)

`~x` flips every bit, which is equivalent to `-x - 1` for integers.

```python
print(~5)
print(~0)
print(~-1)
```

## 3. Left shift (<<) and right shift (>>)

Shifting left multiplies by `2` per position; shifting right divides by `2` (floor) per position.

```python
print(1 << 3)    # 1 * 2**3 = 8
print(8 >> 2)    # 8 // 2**2 = 2
print(5 << 1)    # 5 * 2 = 10
```

## 4. Practical use: flags

Bitwise operators are often used to store several boolean options compactly in a single integer.

```python
READ    = 0b001   # 1
WRITE   = 0b010   # 2
EXECUTE = 0b100   # 4

# Grant read + write permissions
permissions = READ | WRITE
print('permissions:', bin(permissions))

# Check if WRITE is granted
print('has write:', bool(permissions & WRITE))
print('has execute:', bool(permissions & EXECUTE))

# Add execute permission
permissions |= EXECUTE
print('updated:', bin(permissions))

# Remove write permission
permissions &= ~WRITE
print('after removing write:', bin(permissions))
```
