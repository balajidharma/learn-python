# Special Operators

Python has two extra categories of operators that don't fit the arithmetic/comparison/logical/bitwise groups:

| Operator | Category | Meaning |
|----------|----------|---------|
| `in` / `not in` | Membership | Test whether a value exists in a sequence |
| `is` / `is not` | Identity | Test whether two variables point to the **same object** |

## 1. Membership: in / not in

```python
fruits = ['apple', 'banana', 'cherry']

print('banana' in fruits)
print('mango' in fruits)
print('mango' not in fruits)

# Also works on strings and dictionaries
print('a' in 'banana')
print('name' in {'name': 'Alice', 'age': 30})
```

## 2. Identity: is / is not

`is` checks whether two names refer to the **same object in memory**, not just equal values. Use `==` to compare values and `is` to compare identity.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)   # same values -> True
print(a is b)   # different objects -> False
print(a is c)   # c points to the same list as a -> True
```

## 3. is with None

Always use `is` / `is not` (not `==`) when checking for `None`. This is the idiomatic Python style.

```python
value = None

if value is None:
    print('value has not been set')

value = 42
if value is not None:
    print(f'value is {value}')
```

## 4. A common gotcha: is vs ==

Small integers and short strings are sometimes cached by Python, which can make `is` appear to work for equality by coincidence. Never rely on this — always use `==` for value comparisons.

```python
x = 1000
y = 1000
print(x == y)   # True: same value
print(x is y)   # not guaranteed to be True — don't rely on it for large ints
```
