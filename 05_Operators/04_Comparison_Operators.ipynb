# Comparison Operators

Comparison operators compare two values and always return a `bool` (`True` or `False`). They're the building blocks of conditions in `if` statements and loops.

| Operator | Meaning |
|----------|---------|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

## 1. Basic comparisons

```python
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= 10)
print(a <= 5)
```

## 2. Comparing strings

Strings are compared lexicographically (character by character, based on Unicode code points).

```python
print('apple' == 'apple')
print('apple' == 'Apple')   # case-sensitive
print('apple' < 'banana')   # 'a' comes before 'b'
print('Zebra' < 'apple')    # uppercase letters sort before lowercase
```

## 3. Comparing different numeric types

Python compares `int`, `float`, and `bool` by value, not by type.

```python
print(5 == 5.0)
print(1 == True)
print(0 == False)
print(5 == '5')   # different types (int vs str) are never equal
```

## 4. Chained comparisons

Python lets you chain comparisons, which reads naturally like math notation.

```python
age = 25
print(18 <= age <= 65)     # equivalent to: 18 <= age and age <= 65

x = 5
print(1 < x < 10 < 20)
```

## 5. Using comparisons in conditions

```python
temperature = 28

if temperature >= 30:
    print('Hot')
elif temperature >= 20:
    print('Warm')
else:
    print('Cool')
```
