# Logical Operators

Logical operators combine or invert boolean expressions.

| Operator | Meaning |
|----------|---------|
| `and` | `True` if **both** sides are `True` |
| `or` | `True` if **at least one** side is `True` |
| `not` | Inverts a boolean value |

## 1. and

```python
age = 25
has_license = True

print(age >= 18 and has_license)
print(age >= 18 and False)
```

## 2. or

```python
is_weekend = False
is_holiday = True

print(is_weekend or is_holiday)
print(False or False)
```

## 3. not

```python
is_raining = False
print(not is_raining)
print(not True)
```

## 4. Short-circuit evaluation

`and` stops as soon as it finds a falsy value; `or` stops as soon as it finds a truthy value. The remaining expression is never evaluated — useful for avoiding errors.

```python
def risky_check():
    print('risky_check() was called')
    return True

print(False and risky_check())   # risky_check() never runs
print(True or risky_check())     # risky_check() never runs
```

## 5. Combining multiple conditions

```python
username = 'admin'
password = 'secret123'
is_active = True

if username == 'admin' and password == 'secret123' and is_active:
    print('Access granted')
else:
    print('Access denied')
```

## 6. Operator precedence: not, and, or

`not` binds tighter than `and`, which binds tighter than `or`. Use parentheses when mixing them to keep the logic clear.

```python
a, b, c = True, False, True

print(a and b or c)          # (a and b) or c
print(a and (b or c))        # explicit grouping changes the result
print(not a and b)           # (not a) and b
```
