# Arithmetic Operators

Arithmetic operators perform mathematical calculations like addition, subtraction, multiplication, and division.

| Operator | Name | Example |
|----------|------|---------|
| `+` | Addition | `5 + 2` |
| `-` | Subtraction | `5 - 2` |
| `*` | Multiplication | `5 * 2` |
| `/` | Division (true) | `5 / 2` |
| `//` | Floor Division | `5 // 2` |
| `%` | Modulus (remainder) | `5 % 2` |
| `**` | Exponentiation | `5 ** 2` |

## 1. Basic arithmetic

```python
a = 5
b = 2

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

## 2. Floor division vs true division

`/` always returns a `float`. `//` divides and rounds **down** to the nearest whole number.

```python
print(7 / 2)
print(7 // 2)
print(-7 // 2)   # rounds toward negative infinity, not toward zero
```

## 3. Modulus (%)

Returns the remainder of a division. Commonly used to check even/odd numbers or wrap values around a range.

```python
print(10 % 3)
print(9 % 3)

for n in range(1, 6):
    label = 'even' if n % 2 == 0 else 'odd'
    print(n, label)
```

## 4. Exponentiation (**)

Raises a number to a power. Works with negative and fractional exponents too.

```python
print(2 ** 10)
print(9 ** 0.5)   # square root
print(2 ** -2)    # negative exponent -> fraction
```

## 5. Operator precedence

Python follows standard math precedence: parentheses, then exponents, then `*` `/` `//` `%`, then `+` `-`. Use parentheses to make intent clear.

```python
result = 2 + 3 * 4
print(result)          # multiplication happens first

result = (2 + 3) * 4
print(result)          # parentheses change the order

result = 2 ** 3 ** 2   # exponentiation is right-associative: 2 ** (3 ** 2)
print(result)
```

## 6. Mixing types

Mixing `int` and `float` produces a `float`. Arithmetic also works directly on variables holding computed expressions.

```python
price = 19.99
quantity = 3
total = price * quantity
print(f'Total: {total}')
print(type(total))
```
