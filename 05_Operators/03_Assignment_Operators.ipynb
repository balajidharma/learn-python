# Assignment Operators

Assignment operators store values in variables. Python's compound assignment operators combine an arithmetic (or bitwise) operation with assignment in one step.

| Operator | Equivalent to |
|----------|---------------|
| `=` | `x = value` |
| `+=` | `x = x + value` |
| `-=` | `x = x - value` |
| `*=` | `x = x * value` |
| `/=` | `x = x / value` |
| `//=` | `x = x // value` |
| `%=` | `x = x % value` |
| `**=` | `x = x ** value` |
| `&=` `\|=` `^=` `>>=` `<<=` | bitwise equivalents |

## 1. Basic assignment

```python
x = 10
print(x)
```

## 2. Compound arithmetic assignment

```python
score = 50
score += 10   # score = score + 10
print(score)

score -= 5    # score = score - 5
print(score)

score *= 2    # score = score * 2
print(score)

score /= 4    # score = score / 4
print(score)
```

## 3. Floor division, modulus, and power assignment

```python
n = 17
n //= 5
print(n)

n = 17
n %= 5
print(n)

n = 2
n **= 5
print(n)
```

## 4. Bitwise assignment operators

These combine a bitwise operation with assignment (see the [Bitwise Operators](06_Bitwise_Operators.ipynb) page for details on what each bitwise operator does).

```python
flags = 0b1010
flags &= 0b0110
print(bin(flags))

flags = 0b1010
flags |= 0b0101
print(bin(flags))

flags = 0b1010
flags ^= 0b1111
print(bin(flags))

flags = 1
flags <<= 3
print(flags)

flags = 16
flags >>= 2
print(flags)
```

## 5. The walrus operator (`:=`)

Introduced in Python 3.8, the walrus operator assigns a value **and** returns it in a single expression — handy inside `while` loops or comprehensions.

```python
data = [1, 2, 3, 4, 5, 6, 7, 8]

# Without walrus: compute length twice
if len(data) > 5:
    print(f'List is long: {len(data)} items')

# With walrus: compute once, reuse the result
if (n := len(data)) > 5:
    print(f'List is long: {n} items')
```
