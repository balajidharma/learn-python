# Basic Syntax
Python uses indentation to define code blocks, making the code easy to read.

**Key Points:**
- Statements end with a newline, not a semicolon.
- Indentation (usually 4 spaces) is required for blocks (e.g., after `if`, `for`, `def`).
- Comments start with `#`.
- Python is case-sensitive.

```python
# print function
print("Hello, World!")
```

```python
# Example for using Indentation
if True:
   print ("True")
else:
   print ("False")
```

```python
# String Concatenation
name  = 'Balaji'
age = 25
print('String Concatenation: Your name is ' + name.upper() + ' and your are', age, 'years old')
```

```python
# format
result = "Format: Your name is {name} and your age {age} years old".format(name = name.upper(), age = age)
print(result)
```

```python
# F-Strings
print(f"F-Strings: Your name is {name.upper()} and your age {age} years old")
```

```python
# Slicing Strings
print("Slicing Strings", name[0:3], name[3:], name[0:10])
```
