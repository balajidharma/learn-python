## User Input
Python uses the `input()` function to get input from the user. The input is always received as a string. You can convert it to other types (like int or float) as needed.

**Syntax:**
```python
variable = input('Prompt message')
```

**Type Conversion Example:**
```python
age = int(input('Enter your age: '))
```

**Note:** Always handle possible conversion errors when converting input to numbers.

---
### Example: Getting User Input

````python
# Example 1: Simple string input
name = input('Enter your name: ')
print('Hello,', name)
````

````python
# Example 2: Integer input with type conversion
age = int(input('Enter your age: '))
print('You are', age, 'years old')
````

````python
# Example 3: Handling invalid input with try-except
try:
    number = float(input('Enter a number: '))
    print('You entered:', number)
except ValueError:
    print('That was not a valid number!')
````
