# Tuples

A tuple is an **ordered**, **immutable** collection — once created, its items can't be changed, added, or removed. Tuples are created with parentheses `()` (the parentheses are actually optional; the commas are what matter).

**Key Points:**
- Ordered — items keep their position
- Immutable — cannot be modified after creation
- Allow duplicates
- Faster and safer than lists when the data shouldn't change

## 1. Creating tuples

```python
empty = ()
single = (42,)          # trailing comma is required for a 1-item tuple
numbers = (1, 2, 3, 4, 5)
no_parens = 1, 2, 3      # parentheses are optional

print(empty)
print(single, type(single))
print(numbers)
print(no_parens, type(no_parens))
```

## 2. Tuples are immutable

Once created, a tuple's items cannot be reassigned, added, or removed.

```python
point = (3, 4)

try:
    point[0] = 10
except TypeError as e:
    print(f'TypeError: {e}')

# You CAN create a new tuple based on an old one
new_point = point + (5,)
print(new_point)
```

## 3. Indexing and slicing

Works exactly like lists and strings.

```python
colors = ('red', 'green', 'blue', 'yellow', 'purple')

print(colors[0])
print(colors[-1])
print(colors[1:3])
print(colors[::-1])
```

## 4. Tuple packing and unpacking

'Packing' groups values into a tuple; 'unpacking' spreads a tuple's values back into separate variables.

```python
# Packing
point = 3, 4, 5
print(point)

# Unpacking
x, y, z = point
print(x, y, z)

# Swapping values using tuple unpacking
a, b = 1, 2
a, b = b, a
print(a, b)

# Star-unpacking to grab the rest
first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)
```

## 5. Tuple methods

Because tuples are immutable, they only have two methods: `count()` and `index()`.

```python
numbers = (1, 2, 2, 3, 2, 4)

print(numbers.count(2))    # how many times 2 appears
print(numbers.index(3))    # position of the first 3
```

## 6. Nested tuples

```python
nested = (1, (2, 3), (4, (5, 6)))

print(nested[1])
print(nested[2][1])
print(nested[2][1][0])
```

## 7. Tuples vs lists — when to use which

| | Tuple | List |
|---|-------|------|
| Mutable? | No | Yes |
| Syntax | `(1, 2, 3)` | `[1, 2, 3]` |
| Typical use | Fixed records (coordinates, RGB values, database rows) | Collections that grow, shrink, or change |
| Performance | Slightly faster, less memory | Slightly slower |
| Can be a dict key? | Yes (if contents are hashable) | No |

```python
# Tuples can be used as dictionary keys because they're immutable/hashable
locations = {
    (40.7128, -74.0060): 'New York',
    (51.5074, -0.1278): 'London',
}
print(locations[(40.7128, -74.0060)])

# Lists cannot be dict keys
try:
    bad = {[1, 2]: 'oops'}
except TypeError as e:
    print(f'TypeError: {e}')
```

## 8. Named tuples

`collections.namedtuple` creates tuple subclasses with named fields — you get the immutability of a tuple with the readability of attribute access.

```python
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])

p = Point(3, 4)
print(p)
print(p.x, p.y)
print(p[0], p[1])   # still works like a regular tuple

x, y = p            # still unpacks like a regular tuple
print(x, y)
```
