# Sets

A set is an **unordered** collection of **unique** items — duplicates are automatically removed. Sets are created with curly braces `{}` or the `set()` function.

**Key Points:**
- Unordered — items have no fixed position or index
- No duplicates — adding an existing item has no effect
- Mutable — you can add or remove items, but items themselves must be immutable (hashable)
- Optimized for fast membership testing (`in`) and set algebra (union, intersection, ...)

## 1. Creating sets

```python
fruits = {'apple', 'banana', 'cherry'}
print(fruits, type(fruits))

# Duplicates are automatically removed
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)

# Empty set MUST use set(), not {} (that creates an empty dict)
empty_set = set()
empty_dict = {}
print(type(empty_set), type(empty_dict))

# set() can build a set from any iterable
letters = set('mississippi')
print(letters)
```

## 2. Sets have no order or index

You cannot access set items by position — there's no `my_set[0]`.

```python
colors = {'red', 'green', 'blue'}

try:
    colors[0]
except TypeError as e:
    print(f'TypeError: {e}')

print('red' in colors)   # membership testing is what sets are great at
```

## 3. Adding and removing items

```python
fruits = {'apple', 'banana'}

fruits.add('cherry')            # add a single item
print(fruits)

fruits.update(['date', 'fig'])  # add multiple items from any iterable
print(fruits)

fruits.remove('banana')         # raises KeyError if not present
print(fruits)

fruits.discard('mango')         # no error even if not present
print(fruits)

popped = fruits.pop()            # removes an arbitrary item
print('popped:', popped)

fruits.clear()
print(fruits)
```

## 4. Set operations: union, intersection, difference

Sets support mathematical set algebra, either via methods or operators.

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print('union:', a | b, '==', a.union(b))
print('intersection:', a & b, '==', a.intersection(b))
print('difference (a - b):', a - b, '==', a.difference(b))
print('symmetric difference:', a ^ b, '==', a.symmetric_difference(b))
```

## 5. Subsets and supersets

```python
a = {1, 2}
b = {1, 2, 3, 4}

print(a.issubset(b))       # is every item of a in b?
print(b.issuperset(a))     # does b contain all of a?
print(a <= b)              # same as issubset
print({1, 2} & {5, 6} == set())   # no overlap -> disjoint
print({1, 2}.isdisjoint({5, 6}))
```

## 6. Removing duplicates from a list

One of the most common practical uses of a set: quickly de-duplicating a list. Note this does not preserve the original order.

```python
numbers = [3, 1, 2, 3, 4, 1, 5, 2]
unique_numbers = list(set(numbers))
print(sorted(unique_numbers))

# Order-preserving de-duplication needs a different trick
seen = set()
ordered_unique = []
for n in numbers:
    if n not in seen:
        seen.add(n)
        ordered_unique.append(n)
print(ordered_unique)
```

## 7. Set comprehensions

```python
words = ['cat', 'dog', 'elephant', 'ant', 'ox', 'bee']
lengths = {len(w) for w in words}
print(sorted(lengths))

numbers = [1, 2, 2, 3, 3, 3, 4]
print({x ** 2 for x in numbers})
```

## 8. Frozensets

A `frozenset` is an **immutable** version of a set. Because it's immutable, it can be used as a dictionary key or stored inside another set — regular sets cannot.

```python
frozen = frozenset([1, 2, 3])
print(frozen)

try:
    frozen.add(4)
except AttributeError as e:
    print(f'AttributeError: {e}')

# A frozenset can be used as a dict key; a regular set cannot
cache = {frozen: 'cached result'}
print(cache[frozenset([1, 2, 3])])
```
