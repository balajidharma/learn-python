# Lists

A list is an **ordered**, **mutable** collection that can hold items of any type — including duplicates and a mix of types. Lists are created with square brackets `[]`.

**Key Points:**
- Ordered — items keep the position you put them in
- Mutable — you can change, add, or remove items after creation
- Allow duplicates
- Can hold mixed types, including other lists

## 1. Creating lists

```python
empty = []
numbers = [1, 2, 3, 4, 5]
mixed = ['Alice', 30, True, 3.14]

print(empty)
print(numbers)
print(mixed)
print(type(numbers))
```

## 2. Indexing and slicing

Lists are indexed starting at `0`. Negative indexes count from the end. Slicing uses `[start:stop:step]`, same as strings.

```python
fruits = ['apple', 'banana', 'cherry', 'date', 'elderberry']

print(fruits[0])       # first item
print(fruits[-1])      # last item
print(fruits[1:3])     # slice
print(fruits[:2])      # first two
print(fruits[::-1])    # reversed
```

## 3. Modifying lists

Unlike strings, lists can be changed in place — this is what makes them **mutable**.

```python
fruits = ['apple', 'banana', 'cherry']

fruits[1] = 'blueberry'   # replace an item
print(fruits)

fruits.append('date')          # add to the end
print(fruits)

fruits.insert(1, 'apricot')    # insert at a specific position
print(fruits)

fruits.remove('cherry')        # remove by value
print(fruits)

popped = fruits.pop()           # remove and return the last item
print(popped, fruits)

del fruits[0]                   # remove by index
print(fruits)
```

## 4. Useful list methods

```python
numbers = [4, 2, 8, 1, 9, 2]

print(len(numbers))
print(numbers.count(2))     # how many times 2 appears
print(numbers.index(8))     # position of first 8

numbers.sort()               # sort in place, ascending
print(numbers)

numbers.sort(reverse=True)   # descending
print(numbers)

numbers.reverse()             # reverse the current order
print(numbers)

more = [10, 20]
numbers.extend(more)          # append all items from another list
print(numbers)
```

## 5. Membership, concatenation, and repetition

```python
fruits = ['apple', 'banana', 'cherry']

print('banana' in fruits)
print('mango' not in fruits)

combined = fruits + ['date', 'fig']
print(combined)

repeated = [0] * 5
print(repeated)
```

## 6. Nested lists

A list can contain other lists — handy for representing grids or tables.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

print(matrix[0])         # first row
print(matrix[1][2])      # row 1, column 2 -> 6

for row in matrix:
    print(row)
```

## 7. List comprehensions

A concise way to build a new list from an existing iterable. (See the [Comprehensions](../08_Conditionals/08_Comprehensions.ipynb) page for a deeper dive.)

```python
numbers = [1, 2, 3, 4, 5, 6]

squares = [n ** 2 for n in numbers]
print(squares)

evens = [n for n in numbers if n % 2 == 0]
print(evens)
```

## 8. Copying a list — a common pitfall

Assigning a list to a new variable does **not** copy it — both names point to the same list. Use `.copy()`, `list()`, or a slice `[:]` to make an independent copy.

```python
original = [1, 2, 3]
alias = original          # NOT a copy — same list
alias.append(4)
print('original:', original)   # also changed!

original = [1, 2, 3]
copy = original.copy()    # a real, independent copy
copy.append(4)
print('original:', original)
print('copy:', copy)
```

## 9. Sorting with a custom key

`sorted()` returns a new sorted list without changing the original; `key=` controls what's compared.

```python
words = ['banana', 'kiwi', 'apple', 'fig']

by_length = sorted(words, key=len)
print(by_length)

people = [('Alice', 30), ('Bob', 25), ('Charlie', 35)]
by_age = sorted(people, key=lambda person: person[1])
print(by_age)
```
