# Loops

Loops let you repeat a block of code multiple times instead of writing it out by hand. Python has two loop keywords:

| Loop | Use when |
|------|----------|
| `while` | You don't know in advance how many times to repeat — you repeat **while** a condition holds |
| `for` | You're iterating over a known sequence or a fixed number of times |

> **Note:** Deeper dives into `while`, `for`, `break`/`continue`/`pass`, and comprehensions live in the [Conditionals](../08_Conditionals/01_Conditionals.ipynb) section — this page is a self-contained overview covering everything you need to get started with loops.

## 1. while loop

Repeats as long as its condition is `True`. Make sure something inside the loop eventually makes the condition `False`, or you'll get an infinite loop.

```python
count = 1
while count <= 5:
    print(count, end=' ')
    count += 1
print()
```

## 2. for loop

Iterates over the items of any sequence (list, string, range, tuple, dict, ...) in order.

```python
fruits = ['apple', 'banana', 'cherry']
for fruit in fruits:
    print(fruit)
```

## 3. range()

`range(stop)`, `range(start, stop)`, and `range(start, stop, step)` generate sequences of numbers — commonly used to repeat something a fixed number of times.

```python
for i in range(5):
    print(i, end=' ')
print()

for i in range(2, 10, 2):
    print(i, end=' ')
print()

for i in range(10, 0, -1):
    print(i, end=' ')
print()
```

## 4. break, continue, and pass

- `break` exits the loop immediately
- `continue` skips the rest of the current iteration and moves to the next one
- `pass` does nothing — a placeholder where a statement is syntactically required

```python
# break: stop as soon as we find a negative number
for n in [3, 7, -4, 9]:
    if n < 0:
        print('Found a negative number, stopping.')
        break
    print(n)

print('---')

# continue: skip even numbers
for n in range(1, 8):
    if n % 2 == 0:
        continue
    print(n, end=' ')
print()
```

## 5. Nested loops

A loop inside another loop. The inner loop completes all its iterations for each single iteration of the outer loop.

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(f'{i}x{j}={i*j}', end='  ')
    print()
```

## 6. enumerate() — index + value

Use `enumerate()` when you need both the position and the value while looping.

```python
languages = ['Python', 'JavaScript', 'Rust']
for index, lang in enumerate(languages, start=1):
    print(f'{index}. {lang}')
```

## 7. zip() — looping over multiple sequences together

```python
names = ['Alice', 'Bob', 'Charlie']
scores = [92, 85, 78]
for name, score in zip(names, scores):
    print(f'{name}: {score}')
```

## 8. Looping over a dictionary

```python
person = {'name': 'Alice', 'age': 30, 'city': 'NYC'}
for key, value in person.items():
    print(f'{key}: {value}')
```

## 9. The loop else clause

The `else` block on a loop runs only if the loop finished **without** hitting a `break`. It's handy for "search and report if not found" patterns.

```python
target = 11
numbers = [2, 4, 6, 8, 10]

for n in numbers:
    if n == target:
        print(f'Found {target}!')
        break
else:
    print(f'{target} was not found in the list')
```

## 10. Building a result inside a loop

A common pattern: start with an empty container, then grow it inside the loop.

```python
squares = []
for n in range(1, 6):
    squares.append(n ** 2)
print(squares)

total = 0
for n in squares:
    total += n
print(f'Sum of squares: {total}')
```
