# Introduction to Python Collections

Python has several built-in ways to store groups of data. These are called collections. The main types are:

- **List**: A list is like a container that keeps items in order. You can change the items, and you can have the same item more than once.
- **Tuple**: A tuple is like a list, but you cannot change its items after you create it. It also keeps items in order and allows duplicates.
- **Set**: A set is a group of unique items. It does not keep any order, and you cannot have the same item twice. You can add or remove items, but you cannot change them directly.
- **Dictionary**: A dictionary stores data in pairs: a key and a value. Each key must be unique. Dictionaries remember the order you add items (in Python 3.7 and above), and you can change the values.

> **Note:**
> - You can add or remove items from a set, but you cannot change the items themselves.
> - Dictionaries keep the order of items as you add them (in Python 3.7+).
