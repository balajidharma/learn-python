# Basic Syntax
Python uses indentation to define code blocks, making the code easy to read.

**Key Points:**
- Statements end with a newline, not a semicolon.
- Indentation (usually 4 spaces) is required for blocks (e.g., after `if`, `for`, `def`).
- Comments start with `#`.
- Python is case-sensitive.

```python
# print function
print("Hello, World!")
```

```python
# Example for using Indentation
if True:
   print ("True")
else:
   print ("False")
```

```python
# Single line comment
print("This is a single line comment example.")

# Multi-line comment
"""This is a multi-line comment example.
It can span multiple lines."""
print("This is a multi-line comment example.")

# multiple single line comments
# This is the first single line comment
# This is the second single line comment
```

```python
# Python case sensitivity
print("Hello, World!")
print("hello, world!")

# Example: Variable names
age = 25
Age = 30

print(age)  # Output: 25
print(Age)  # Output: 30
```
