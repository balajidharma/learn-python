# Strings

A string is a sequence of characters used to represent text. Strings in Python are **immutable** — once created, their contents can't be changed in place.

**Key Points:**
- Create with single `'...'`, double `"..."`, or triple `'''...'''` / `"""..."""` quotes
- Indexable and sliceable like a list
- Immutable — string methods return a *new* string
- Support `+` (concatenation) and `*` (repetition)

## 1. Creating strings

```python
single = 'Hello'
double = "World"
multiline = '''This is
a multi-line
string.'''

print(single, double)
print(multiline)
```

## 2. Indexing and slicing

Strings are indexed starting at `0`. Negative indexes count from the end. Slicing uses `[start:stop:step]`.

```python
text = 'Python'

print(text[0])      # first character
print(text[-1])     # last character
print(text[0:3])    # slice: first 3 characters
print(text[2:])     # from index 2 to the end
print(text[::-1])   # reversed string
```

## 3. Concatenation and repetition

```python
first = 'Learn'
second = 'Python'

print(first + ' ' + second)
print('ab' * 3)
```

## 4. String length

```python
text = 'Hello, Python!'
print(len(text))
```

## 5. Common string methods

Strings are immutable, so every method below **returns a new string** instead of modifying the original.

```python
text = '  Hello, Python!  '

print(text.upper())
print(text.lower())
print(text.strip())
print(text.strip().replace('Python', 'World'))
print(text.strip().startswith('Hello'))
print(text.strip().endswith('!'))
print(text.strip().find('Python'))
```

## 6. Splitting and joining

```python
sentence = 'Python is fun to learn'
words = sentence.split()
print(words)

csv_line = 'apple,banana,cherry'
print(csv_line.split(','))

joined = '-'.join(words)
print(joined)
```

## 7. f-strings (formatted string literals)

The modern, preferred way to build strings with embedded values. Prefix the string with `f` and place expressions inside `{}`.

```python
name = 'Alice'
age = 30

print(f'{name} is {age} years old')
print(f'Next year, {name} will be {age + 1}')

pi = 3.14159265
print(f'Pi rounded to 2 decimals: {pi:.2f}')
```

## 8. Other formatting styles

You may also see `str.format()` and `%`-formatting in older code.

```python
name = 'Bob'
age = 25

print('{} is {} years old'.format(name, age))
print('%s is %d years old' % (name, age))
```

## 9. Escape characters

Use a backslash to include special characters inside a string.

```python
print('Line1\nLine2')      # newline
print('Column1\tColumn2')  # tab
print('She said \'hi\'')   # escaped quote
print("Path: C:\\Users\\Bob") # escaped backslash
```

## 10. Strings are immutable

```python
text = 'Python'
new_text = text.replace('P', 'C')

print(text)       # unchanged
print(new_text)   # a brand-new string

try:
    text[0] = 'C'   # this raises an error
except TypeError as e:
    print(f'TypeError: {e}')
```

## 11. Checking string content

```python
values = ['123', 'abc', 'abc123', '  ', 'Hello World']
for v in values:
    print(f'{v!r:<15} isdigit={v.isdigit()!s:<6} isalpha={v.isalpha()!s:<6} isspace={v.isspace()}')
```
