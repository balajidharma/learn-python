# Dictionaries

A dictionary stores data as **key-value pairs**. Each key must be unique and is used to look up its associated value — like a real-world dictionary where a word (key) maps to its definition (value). Dictionaries are created with curly braces `{key: value, ...}`.

**Key Points:**
- Keys must be unique and immutable (strings, numbers, tuples, ...)
- Values can be anything, including lists, other dicts, or duplicates
- Mutable — you can add, change, or remove pairs
- Ordered by insertion since Python 3.7+

## 1. Creating dictionaries

```python
empty = {}
person = {'name': 'Alice', 'age': 30, 'city': 'NYC'}
from_pairs = dict([('a', 1), ('b', 2)])
from_kwargs = dict(x=1, y=2)

print(empty)
print(person)
print(from_pairs)
print(from_kwargs)
print(type(person))
```

## 2. Accessing values

```python
person = {'name': 'Alice', 'age': 30}

print(person['name'])          # KeyError if the key doesn't exist
print(person.get('age'))       # safer: returns None if missing
print(person.get('email'))
print(person.get('email', 'not provided'))   # custom default

try:
    print(person['email'])
except KeyError as e:
    print(f'KeyError: {e}')
```

## 3. Adding, updating, and removing items

```python
person = {'name': 'Alice', 'age': 30}

person['email'] = 'alice@example.com'   # add a new key
print(person)

person['age'] = 31                       # update an existing key
print(person)

person.update({'age': 32, 'city': 'NYC'})   # update/add several at once
print(person)

del person['city']                       # remove by key
print(person)

removed = person.pop('email')            # remove and return the value
print(removed, person)

last = person.popitem()                  # remove and return the last-inserted pair
print(last, person)
```

## 4. Checking for keys

Use `in` to test for a key's presence — checking `in` on a dict looks at keys, not values.

```python
person = {'name': 'Alice', 'age': 30}

print('name' in person)
print('email' in person)
print('email' not in person)
```

## 5. keys(), values(), and items()

These give you view objects over the dictionary's keys, values, and key-value pairs — commonly used for looping.

```python
person = {'name': 'Alice', 'age': 30, 'city': 'NYC'}

print(list(person.keys()))
print(list(person.values()))
print(list(person.items()))

for key in person:            # looping directly iterates over keys
    print(key, '->', person[key])
```

## 6. Looping with .items()

```python
scores = {'Alice': 92, 'Bob': 74, 'Charlie': 85}

for name, score in scores.items():
    status = 'passed' if score >= 80 else 'failed'
    print(f'{name}: {score} ({status})')
```

## 7. setdefault() — get or insert in one step

`setdefault(key, default)` returns the value for `key` if it exists; otherwise it inserts `key` with `default` and returns that.

```python
word_counts = {}
text = ['apple', 'banana', 'apple', 'cherry', 'banana', 'apple']

for word in text:
    word_counts[word] = word_counts.setdefault(word, 0) + 1

print(word_counts)
```

## 8. Nested dictionaries

Dictionary values can themselves be dictionaries — useful for representing structured records.

```python
users = {
    'alice': {'age': 30, 'email': 'alice@example.com'},
    'bob':   {'age': 25, 'email': 'bob@example.com'},
}

print(users['alice']['email'])

for username, info in users.items():
    print(f"{username}: age={info['age']}, email={info['email']}")
```

## 9. Dictionary comprehensions

```python
squares = {x: x ** 2 for x in range(1, 6)}
print(squares)

scores = {'Alice': 92, 'Bob': 74, 'Charlie': 85, 'Diana': 68}
passed = {name: score for name, score in scores.items() if score >= 80}
print(passed)
```

## 10. Merging dictionaries

Python 3.9+ supports the `|` merge operator. Older code typically uses `.update()` or `{**a, **b}` unpacking.

```python
defaults = {'theme': 'light', 'font_size': 12}
user_prefs = {'font_size': 16, 'language': 'en'}

merged = defaults | user_prefs        # Python 3.9+: right side wins on conflicts
print(merged)

merged_unpack = {**defaults, **user_prefs}   # works on older Python too
print(merged_unpack)
```
