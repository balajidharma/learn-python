# Typecasting

Typecasting (type conversion) means converting a value from one data type to another. Python does this in two ways:

- **Implicit conversion** — Python converts automatically, without you asking, when it's safe to do so (no data is lost).
- **Explicit conversion** — you convert manually using built-in functions like `int()`, `float()`, `str()`, and `bool()`.

## 1. Implicit type conversion

When an expression mixes types, Python automatically promotes the "smaller" type to the "larger" one to avoid losing information — e.g. `int` + `float` becomes `float`.

```python
a = 5        # int
b = 2.5      # float

c = a + b    # int is implicitly converted to float
print(c)
print(type(c))

# Booleans are treated as 0 / 1 in numeric contexts
result = True + 3
print(result)
print(type(result))
```

## 2. Explicit conversion: int()

Converts a float (truncates the decimal part), a numeric string, or a bool to an integer.

```python
print(int(7.9))       # truncates toward zero, does NOT round
print(int(-7.9))
print(int('42'))
print(int(True))
print(int(False))

try:
    int('abc')
except ValueError as e:
    print(f'ValueError: {e}')
```

## 3. Explicit conversion: float()

Converts an int or a numeric string to a float.

```python
print(float(5))
print(float('3.14'))
print(float('10'))

try:
    float('not a number')
except ValueError as e:
    print(f'ValueError: {e}')
```

## 4. Explicit conversion: str()

Converts almost any value into its string representation — useful for concatenating with text.

```python
age = 30
print('Age: ' + str(age))

pi = 3.14159
print('Pi: ' + str(pi))

is_valid = True
print('Valid: ' + str(is_valid))

print(str([1, 2, 3]))
```

## 5. Explicit conversion: bool()

Every value in Python is either truthy or falsy. `bool()` converts a value to `True` or `False` based on that.

**Falsy:** `0`, `0.0`, `''`, `[]`, `{}`, `None`, `False`
**Truthy:** everything else

```python
print(bool(0))
print(bool(1))
print(bool(''))
print(bool('hello'))
print(bool([]))
print(bool([1, 2]))
print(bool(None))
```

## 6. Converting between collections

`list()`, `tuple()`, and `set()` can convert any iterable into that collection type.

```python
text = 'hello'
print(list(text))     # string -> list of characters

numbers = [1, 2, 2, 3, 3, 3]
print(set(numbers))   # list -> set (removes duplicates)

pairs = [('a', 1), ('b', 2)]
print(dict(pairs))    # list of pairs -> dict

print(tuple([1, 2, 3]))  # list -> tuple
```

## 7. Handling conversion errors safely

Explicit conversions can fail if the value doesn't make sense in the target type. Wrap risky conversions in `try` / `except` — especially when converting user input.

```python
def safe_int(value, default=0):
    try:
        return int(value)
    except (ValueError, TypeError):
        print(f'Could not convert {value!r} to int, using default {default}')
        return default

print(safe_int('42'))
print(safe_int('not a number'))
print(safe_int(None))
```

## 8. Practical example: converting user input

`input()` always returns a `str`. Convert it explicitly before doing math with it.

```python
raw_input_value = '25'   # simulates what input() would return

age = int(raw_input_value)
print(f'In 10 years you will be {age + 10}')
print(type(age))
```
